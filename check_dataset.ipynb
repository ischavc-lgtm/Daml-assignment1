```python
import os

import pandas as pd

data = pd.read_csv("cleaned-data/cleaned-data.csv")

```

```python
#Checks length of dataset, should be 41
print("Number of items:", len(data))
```

```python
#Checks if ID's are unique
print("Repeated IDs:", data["ID"].duplicated().sum())
```

```python
#Counts missing calues in each column, there should be zero missing values
data.isna().sum()
```

```python
#lists all different ratings and counts
data["Rating"].value_counts()
```

```python
#calculates average rating
average_rating = data["Rating"].mean()
print("Average rating is:", average_rating)
```

```python
#list of decades that films were made in 
decades = []
for year in data["Year"].dropna():
    decades.append(int(year) // 10 * 10)

pd.Series(decades).value_counts().sort_index()
```
